### Solution 1: Basic TF-IDF

````python
ex1_vectorizer = TfidfVectorizer()
ex1_tfidf_matrix = ex1_vectorizer.fit_transform(documents)

print(f"Shape: {ex1_tfidf_matrix.shape}")
print("\nRows = documents (5)")
print("Columns = unique words in vocabulary")
````

### Solution 2: Vocabulary

````python
ex2_vocab = ex1_vectorizer.get_feature_names_out()
print(f"Vocabulary: {list(ex2_vocab)}")

ex2_neural_index = list(ex2_vocab).index("neural")
print(f"\nIndex of 'neural': {ex2_neural_index}")

ex2_word_at_5 = ex2_vocab[5]
print(f"Word at index 5: {ex2_word_at_5}")
````

### Solution 3: Interpreting Scores

````python
ex3_df = pd.DataFrame(
    ex1_tfidf_matrix.toarray(),
    columns=ex2_vocab,
    index=[f"doc_{i}" for i in range(5)]
)
print(ex3_df.round(3))

print(f"\nDoc 1 highest score: {ex3_df.iloc[1].idxmax()} = {ex3_df.iloc[1].max():.3f}")
print(f"\nWords with 0 score in doc 1: {list(ex3_df.columns[ex3_df.iloc[1] == 0])}")
print("\nWords have 0 score because they don't appear in that document.")
````

### Solution 4: Count vs TF-IDF

````python
ex4_count_vec = CountVectorizer()
ex4_count_matrix = ex4_count_vec.fit_transform(documents)

the_idx_count = list(ex4_count_vec.get_feature_names_out()).index("the")
the_idx_tfidf = list(ex2_vocab).index("the")

print(f"'the' in CountVectorizer: {ex4_count_matrix.toarray()[:, the_idx_count]}")
print(f"'the' in TF-IDF: {ex1_tfidf_matrix.toarray()[:, the_idx_tfidf].round(3)}")
print("\nTF-IDF downweights 'the' because it appears in many documents (low IDF).")
print("Common words are less useful for distinguishing documents.")
````

### Solution 5: Custom Parameters

````python
ex5_vectorizer = TfidfVectorizer(max_features=10, stop_words='english')
ex5_matrix = ex5_vectorizer.fit_transform(documents)

print(f"Original vocab size: {len(ex2_vocab)}")
print(f"New vocab size: {len(ex5_vectorizer.get_feature_names_out())}")
print(f"\nNew vocab: {list(ex5_vectorizer.get_feature_names_out())}")

removed = set(ex2_vocab) - set(ex5_vectorizer.get_feature_names_out())
print(f"\nRemoved words: {removed}")
````

### Solution 6: N-grams

````python
ex6_vectorizer = TfidfVectorizer(ngram_range=(1, 2))
ex6_matrix = ex6_vectorizer.fit_transform(documents)
ex6_vocab = ex6_vectorizer.get_feature_names_out()

bigrams = [w for w in ex6_vocab if ' ' in w]
print(f"Bigrams found: {bigrams}")

if 'neural networks' in ex6_vocab:
    nn_idx = list(ex6_vocab).index('neural networks')
    print(f"\n'neural networks' TF-IDF scores:")
    for i, doc in enumerate(documents):
        print(f"  doc_{i}: {ex6_matrix.toarray()[i, nn_idx]:.3f}")
````

### Solution 7: Document Similarity

````python
from sklearn.metrics.pairwise import cosine_similarity

ex7_similarities = cosine_similarity(ex1_tfidf_matrix)

sim_df = pd.DataFrame(
    ex7_similarities,
    index=[f"doc_{i}" for i in range(5)],
    columns=[f"doc_{i}" for i in range(5)]
)
print("Similarity Matrix:")
print(sim_df.round(3))

# Find most similar pair (excluding diagonal)
sim_copy = ex7_similarities.copy()
np.fill_diagonal(sim_copy, 0)
max_idx = np.unravel_index(sim_copy.argmax(), sim_copy.shape)
print(f"\nMost similar: doc_{max_idx[0]} and doc_{max_idx[1]}")
print(f"  '{documents[max_idx[0]]}'")
print(f"  '{documents[max_idx[1]]}'")
````

### Solution 8: Combining Text Fields

````python
ex8_combined = [p["title"] + " " + p["description"] for p in ml_projects]
print("Combined text:")
for i, text in enumerate(ex8_combined):
    print(f"  {i}: {text}")

ex8_vectorizer = TfidfVectorizer()
ex8_matrix = ex8_vectorizer.fit_transform(ex8_combined)
ex8_vocab = ex8_vectorizer.get_feature_names_out()

if 'bert' in ex8_vocab:
    bert_idx = list(ex8_vocab).index('bert')
    bert_scores = ex8_matrix.toarray()[:, bert_idx]
    print(f"\nBERT scores: {bert_scores.round(3)}")
    winner = bert_scores.argmax()
    print(f"Highest BERT score: Project {winner} - {ml_projects[winner]['title']}")
````

### Solution 9: Manual Calculation

````python
doc2_words = documents[1].split()
print(f"Document 2: {documents[1]}")
print(f"Words: {doc2_words}")

# TF
neural_count = doc2_words.count("neural")
tf = neural_count / len(doc2_words)
print(f"\nTF('neural'): {neural_count}/{len(doc2_words)} = {tf:.4f}")

# IDF
docs_with_neural = sum(1 for d in documents if "neural" in d)
idf = np.log(len(documents) / docs_with_neural) + 1
print(f"IDF('neural'): log({len(documents)}/{docs_with_neural}) + 1 = {idf:.4f}")

# TF-IDF
tfidf_manual = tf * idf
print(f"TF-IDF (manual): {tfidf_manual:.4f}")

# Compare to sklearn
neural_idx = list(ex2_vocab).index("neural")
sklearn_value = ex1_tfidf_matrix.toarray()[1, neural_idx]
print(f"\nsklearn TF-IDF: {sklearn_value:.4f}")
print("\nNote: sklearn normalizes by L2 norm of the document vector, so values differ.")
````

### Solution 10: Classification Pipeline

````python
from sklearn.linear_model import LogisticRegression

# Prepare data
combined_text = [p["title"] + " " + p["description"] for p in ml_projects]
labels = ["nlp", "cv", "nlp", "cv", "nlp"]

# Vectorize
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(combined_text)

# Train
clf = LogisticRegression()
clf.fit(X, labels)

# Predict
new_project = "Transformer model for language translation"
new_vector = vectorizer.transform([new_project])
prediction = clf.predict(new_vector)

print(f"New project: '{new_project}'")
print(f"Predicted category: {prediction[0]}")

# Show probabilities
probs = clf.predict_proba(new_vector)[0]
print(f"\nProbabilities:")
for cls, prob in zip(clf.classes_, probs):
    print(f"  {cls}: {prob:.2%}")
````
